```python
import happybase
import itertools

# HBase REST API 엔드포인트 및 테이블 이름 설정
host  = 'kwall7.kweather.co.kr'  # HBase REST API 서버 URL
table_name  = 'tsdb-uid_kw2'  # HBase 테이블 이름

# HBase 연결 설정
connection = happybase.Connection(host)
table = connection.table(table_name)

# Define the row key you want to retrieve
row_key = b'kw-airkorea.26133'
results = table.scan(filter=f"PrefixFilter('{row_key}')")

# 결과 출력
for key, data in results:
    print(f'Row key: {key.decode("utf-8")}')
    for column, timestamp_value in data.items():
        timestamp, value = timestamp_value
        print(f'Column: {column.decode("utf-8")}, Timestamp: {timestamp}, Value: {value}')

# 연결 종료
connection.close()
```

```python
import happybase

# HBase REST API 엔드포인트 및 테이블 이름 설정
host  = 'kwall7.kweather.co.kr'  # HBase REST API 서버 URL
table_name  = 'tsdb-uid_kw2'  # HBase 테이블 이름

# HBase 연결 설정
connection = happybase.Connection(host)
table = connection.table(table_name)

# 삭제할 데이터의 row key
row_key_to_delete = 'test.kma-aq.10002'

# 데이터 삭제
table.delete(row_key_to_delete.encode())

# 데이터 삭제 후 확인
data = table.row(row_key_to_delete.encode())
if not data:
    print(f'Data with row key {row_key_to_delete} has been deleted.')
else:
    print(f'Data with row key {row_key_to_delete} still exists.')
```
